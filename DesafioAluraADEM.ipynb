### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Sumamos Precio + Costo de envío por cada fila
tienda['Ingreso'] = tienda['Precio'] + tienda['Costo de envío']
tienda2['Ingreso'] = tienda2['Precio'] + tienda2['Costo de envío']
tienda3['Ingreso'] = tienda3['Precio'] + tienda3['Costo de envío']
tienda4['Ingreso'] = tienda4['Precio'] + tienda4['Costo de envío']

# Sumamos los ingresos totales por tienda
ingreso_tienda1 = tienda['Ingreso'].sum()
ingreso_tienda2 = tienda2['Ingreso'].sum()
ingreso_tienda3 = tienda3['Ingreso'].sum()
ingreso_tienda4 = tienda4['Ingreso'].sum()

# Mostramos los resultados
print(f"Ingreso total tienda 1: {ingreso_tienda1}")
print(f"Ingreso total tienda 2: {ingreso_tienda2}")
print(f"Ingreso total tienda 3: {ingreso_tienda3}")
print(f"Ingreso total tienda 4: {ingreso_tienda4}")

```

# 2. Ventas por categoría

```python
# Categorías más populares por tienda

ventas_tienda1 = tienda.groupby('Categoría del Producto').size().sort_values(ascending=False)
ventas_tienda2 = tienda2.groupby('Categoría del Producto').size().sort_values(ascending=False)
ventas_tienda3 = tienda3.groupby('Categoría del Producto').size().sort_values(ascending=False)
ventas_tienda4 = tienda4.groupby('Categoría del Producto').size().sort_values(ascending=False)

# Mostrar los resultados
print("Ventas por categoría - Tienda 1:")
print(ventas_tienda1)
print("\nVentas por categoría - Tienda 2:")
print(ventas_tienda2)
print("\nVentas por categoría - Tienda 3:")
print(ventas_tienda3)
print("\nVentas por categoría - Tienda 4:")
print(ventas_tienda4)
```

```python
import pandas as pd

# Combinar todos los DataFrames en uno solo
todas_las_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4])

# Agrupar por categoría y contar ventas
ventas_por_categoria = todas_las_tiendas.groupby('Categoría del Producto').size()

# Ordenar de mayor a menor
ventas_ordenadas = ventas_por_categoria.sort_values(ascending=False)

# Mostrar el top 5
top_5_categorias = ventas_ordenadas.head(5)
print("Top 5 categorías más vendidas entre todas las tiendas:")
print(top_5_categorias)
```

# 3. Calificación promedio de la tienda


```python
# Calificación promedio por tienda
calificacion_promedio_tienda1 = tienda['Calificación'].mean()
calificacion_promedio_tienda2 = tienda2['Calificación'].mean()
calificacion_promedio_tienda3 = tienda3['Calificación'].mean()
calificacion_promedio_tienda4 = tienda4['Calificación'].mean()

# Mostrar los resultados
print(f"Calificación promedio tienda 1: {calificacion_promedio_tienda1:.2f}")
print(f"Calificación promedio tienda 2: {calificacion_promedio_tienda2:.2f}")
print(f"Calificación promedio tienda 3: {calificacion_promedio_tienda3:.2f}")
print(f"Calificación promedio tienda 4: {calificacion_promedio_tienda4:.2f}")
```

```python
import matplotlib.pyplot as plt

# Histograma de calificaciones para cada tienda
tienda['Calificación'].plot(kind='hist', bins=10, alpha=0.7, label='Tienda 1')
tienda2['Calificación'].plot(kind='hist', bins=10, alpha=0.7, label='Tienda 2')
tienda3['Calificación'].plot(kind='hist', bins=10, alpha=0.7, label='Tienda 3')
tienda4['Calificación'].plot(kind='hist', bins=10, alpha=0.7, label='Tienda 4')

plt.title('Distribución de Calificaciones por Tienda')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```

# 4. Productos más y menos vendidos

```python
# Productos más vendidos
productos_mas_vendidos_tienda1 = tienda.groupby('Producto').size().sort_values(ascending=False)
productos_mas_vendidos_tienda2 = tienda2.groupby('Producto').size().sort_values(ascending=False)
productos_mas_vendidos_tienda3 = tienda3.groupby('Producto').size().sort_values(ascending=False)
productos_mas_vendidos_tienda4 = tienda4.groupby('Producto').size().sort_values(ascending=False)

# Productos menos vendidos
productos_menos_vendidos_tienda1 = tienda.groupby('Producto').size().sort_values(ascending=True)
productos_menos_vendidos_tienda2 = tienda2.groupby('Producto').size().sort_values(ascending=True)
productos_menos_vendidos_tienda3 = tienda3.groupby('Producto').size().sort_values(ascending=True)
productos_menos_vendidos_tienda4 = tienda4.groupby('Producto').size().sort_values(ascending=True)

# Mostrar los 5 productos más y menos vendidos de cada tienda
print("Productos más vendidos - Tienda 1:")
print(productos_mas_vendidos_tienda1.head(5))
print("\nProductos menos vendidos - Tienda 1:")
print(productos_menos_vendidos_tienda1.head(5))

print("\nProductos más vendidos - Tienda 2:")
print(productos_mas_vendidos_tienda2.head(5))
print("\nProductos menos vendidos - Tienda 2:")
print(productos_menos_vendidos_tienda2.head(5))

print("\nProductos más vendidos - Tienda 3:")
print(productos_mas_vendidos_tienda3.head(5))
print("\nProductos menos vendidos - Tienda 3:")
print(productos_menos_vendidos_tienda3.head(5))

print("\nProductos más vendidos - Tienda 4:")
print(productos_mas_vendidos_tienda4.head(5))
print("\nProductos menos vendidos - Tienda 4:")
print(productos_menos_vendidos_tienda4.head(5))
```

# 5. Envío promedio por tienda

```python
# Costo de envío promedio por tienda
costo_envio_promedio_tienda1 = tienda['Costo de envío'].mean()
costo_envio_promedio_tienda2 = tienda2['Costo de envío'].mean()
costo_envio_promedio_tienda3 = tienda3['Costo de envío'].mean()
costo_envio_promedio_tienda4 = tienda4['Costo de envío'].mean()

# Mostrar los resultados
print(f"Costo de envío promedio tienda 1: {costo_envio_promedio_tienda1:.2f}")
print(f"Costo de envío promedio tienda 2: {costo_envio_promedio_tienda2:.2f}")
print(f"Costo de envío promedio tienda 3: {costo_envio_promedio_tienda3:.2f}")
print(f"Costo de envío promedio tienda 4: {costo_envio_promedio_tienda4:.2f}")
```

# **7 EXPLICACION FINAL**


El análisis realizado sobre las tiendas, basado en factores clave como ingresos, calificaciones de clientes, distribución de productos y costos de envío, nos permite determinar cuál de las cuatro tiendas es la más adecuada para que el Sr. Juan decida vender. A continuación, se detallan los hallazgos más relevantes de cada tienda y la recomendación final.

1. Ingresos Totales:

Los ingresos totales de cada tienda fueron calculados sumando tanto el precio del producto como el costo de envío. De los datos obtenidos, observamos que:

Tienda 1 presentó el ingreso total más alto, lo que indica que genera mayores ventas en términos absolutos.

Tienda 2 y Tienda 3 tienen ingresos totales significativamente más bajos en comparación con la Tienda 1.

Tienda 4 muestra un ingreso intermedio, por lo que se encuentra en una posición competitiva, pero no es la líder en cuanto a ingresos.

Esto indica que la Tienda 1 tiene el mejor rendimiento en términos de generación de ingresos, mientras que las demás tiendas muestran resultados más modestos.

2. Distribución de Categorías de Productos:

El análisis de la distribución de categorías revela qué tan variadas son las ventas entre las tiendas. La Tienda 1 tiene una gran variedad de productos en varias categorías, lo que podría significar una mayor versatilidad para el Sr. Juan, dependiendo del tipo de productos que quiera ofrecer.

Tienda 1 muestra una distribución más equilibrada entre varias categorías de productos, lo que podría atraer a más tipos de consumidores.

Tienda 2 tiene una categoría dominante, pero esta especialización podría limitar la oferta y la demanda.

Tienda 3 y Tienda 4 también tienen una variedad de productos, pero sus ventas son más concentradas en menos categorías.

El hecho de que la Tienda 1 ofrezca una gama amplia de productos sugiere que puede llegar a un público más diverso, lo que beneficia a un vendedor con productos variados.

3. Calificaciones de Clientes:

Las calificaciones promedio reflejan la satisfacción de los clientes con los productos vendidos. Es un indicador clave para entender la calidad de los productos y la confianza de los consumidores:

Tienda 1 presenta una calificación promedio alta, lo que sugiere que los clientes están satisfechos con los productos y servicios ofrecidos.

Tienda 2 y Tienda 4 tienen calificaciones ligeramente más bajas, lo que podría indicar ciertos aspectos que necesitan mejora, ya sea en la calidad de los productos o en el servicio al cliente.

Tienda 3 tiene una calificación promedio aún más baja, lo que podría ser una señal de que los productos o el servicio al cliente no cumplen con las expectativas de los consumidores.

La Tienda 1 destaca como la opción con la mejor reputación entre los clientes, lo cual es un factor fundamental para la fidelización y el éxito a largo plazo.

4. Costos de Envío:

Los costos de envío también son un factor relevante para determinar qué tan atractivo es vender en cada tienda. Un costo de envío bajo o razonable es importante tanto para los clientes como para los vendedores:

Tienda 1 tiene un costo de envío promedio moderado, lo que indica que los precios son competitivos sin ser excesivamente altos.

Tienda 2 y Tienda 4 tienen costos de envío un poco más altos, lo cual podría impactar en la percepción del precio final para los clientes.

Tienda 3 tiene el costo de envío más bajo, lo que podría ser atractivo para los clientes, pero su bajo rendimiento en términos de ingresos podría indicar que, a pesar del costo bajo, no genera suficientes ventas.

Aunque la Tienda 3 tiene los costos de envío más bajos, la Tienda 1 ofrece un balance adecuado entre ingresos, costos y satisfacción del cliente, lo que la convierte en la opción más atractiva a pesar de los costos de envío ligeramente más altos.

Recomendación:

Basado en los análisis anteriores, recomendamos que el Sr. Juan venda en la Tienda 1. Esta tienda destaca en varios aspectos clave:

Altos ingresos totales, lo que indica un mayor volumen de ventas y una base de clientes estable.

Una distribución equilibrada de categorías de productos, lo que permite atraer a una audiencia diversa.

Calificaciones altas de los clientes, lo que sugiere una buena reputación y confianza en los productos y servicios.

Un costo de envío competitivo, que no es el más bajo pero está en un rango razonable, permitiendo un equilibrio entre costos y satisfacción del cliente.

Aunque la Tienda 3 tiene costos de envío más bajos, su rendimiento general en ingresos y calificación no es tan fuerte como el de la Tienda 1. Por lo tanto, la Tienda 1 representa la mejor oportunidad para que el Sr. Juan logre ventas exitosas y a largo plazo.

Conclusión:

La Tienda 1 es la opción más prometedora para el Sr. Juan, ya que ofrece una combinación de altos ingresos, una amplia variedad de productos, buenas calificaciones de clientes y un costo de envío competitivo. Esto la convierte en el lugar ideal para vender productos y obtener un retorno de inversión positivo.
